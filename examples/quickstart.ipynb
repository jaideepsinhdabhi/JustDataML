# JustDataML quickstart

Train on a DataFrame, look at the results and predict. Needs `pip install "JustDataML[plots]"`.

Run it from the `examples` folder so `Data/` is found.

```python
import pandas as pd
import jdml

df = pd.read_csv('Data/bezdekIris.data')
df.head()
```

```python
model = jdml.fit(df, target='class', feature_selection='importance', plots=True, cv_folds=3)
model
```

```python
model.report()[['Model', 'Score_validation', 'CV_accuracy_mean', 'CV_accuracy_std']]
```

```python
model.features
```

```python
from IPython.display import Image
Image('Output/plots/model_comparison.png')
```

```python
model.predict(df.drop(columns='class').head())
```
